## Mineração

```python
import sys
import os
import joblib

sys.path.append(os.path.abspath(os.path.join('..')))

from sklearn.metrics import accuracy_score
from src.models import split_data, train_random_forest, train_mlp_neural_network
from src.visualization import plot_classification_metrics, plot_confusion_matrix, compare_performances
```

```python
df = joblib.load('../data/processed/dataset_final.pkl')
X_train, X_test, y_train, y_test = split_data(df)
```

```python
# Random Forest
model_rf = train_random_forest(X_train, y_train)
```

```python
# MLP
model_mlp = train_mlp_neural_network(X_train, y_train)
```

```python
# Avaliação do Random Forest
plot_classification_metrics(model_rf, X_test, y_test, "Random Forest")
```

```python
plot_confusion_matrix(model_rf, X_test, y_test, "Random Forest")
```

```python
# Acurácia 
acc_rf = accuracy_score(y_test, model_rf.predict(X_test))
print(f"Acurácia do Random Forest: {acc_rf:.4f}")
```

```python
# Avaliando MLP
plot_classification_metrics(model_mlp, X_test, y_test, "Rede Neural MLP")
```

```python
plot_confusion_matrix(model_mlp, X_test, y_test, "Rede Neural MLP")
```

```python
acc_mlp = accuracy_score(y_test, model_mlp.predict(X_test))
print(f"Acurácia da Rede Neural MLP: {acc_mlp:.4f}")
```

```python
# comparativo das acurácias 
compare_performances(acc_rf, acc_mlp)
```
