## Análise Exploratória de Dados

```python
import sys
import os 

sys.path.append(os.path.abspath(os.path.join('..')))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_raw_data, get_labeled_data
```

```python
df_raw = load_raw_data()
df = get_labeled_data(df_raw)
```

```python
# Detectar outliers 
df['idade'].describe()
```

```python
# Visualização de dados nulos por coluna
plt.figure(figsize=(10, 6)) 
df.isnull().sum().plot(kind='bar', color='skyblue')

plt.title('Contagem de Valores faltantes por atributo')
plt.ylabel('Quantidade de nulos')
plt.xticks(rotation=45, ha='right') 

plt.tight_layout() 

plt.savefig('../reports/figures/ContagemNulos.png', bbox_inches='tight', dpi=300)
plt.show()
```

```python
# Distribuição do Cargo (Variável alvo)
plt.figure(figsize=(1))
sns.countplot(y='cargo', data=df, order=df['cargo'].value_counts().index, palette='viridis')
plt.title('Distribuição da variável alvo: cargo')
plt.xlabel('Número de respondentes')
plt.savefig ('../reports/figures/DistribuicaoCargos.png')
plt.show()
```

```python
# Experiência X Nível de Ensino
plt.figure(figsize=(12, 8))
sns.countplot(y='tempo_experiencia_dados', hue='nivel_ensino', data=df)
plt.title('Experiência na Área vs Nível de Ensino')
plt.legend(title='Escolaridade', bbox_to_anchor=(1, 1))
plt.savefig ('../reports/figures/ExperienciaXEnsino.png')
plt.show()
```
